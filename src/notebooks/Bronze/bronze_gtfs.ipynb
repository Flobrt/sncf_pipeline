```python
import requests
import zipfile
import os
import io
from datetime import datetime
```

```python
dbutils.widgets.text("runid", "2026-05-13_12-43-46")
runid = dbutils.widgets.get("runid")
```

```python
url = "https://eu.ftp.opendatasoft.com/sncf/plandata/Export_OpenData_SNCF_GTFS_NewTripId.zip"

print("Téléchargement en cours...")
response = requests.get(url, stream=True)

print("Extraction en cours...")
with zipfile.ZipFile(io.BytesIO(response.content)) as z:
    for filename in z.namelist():
        print(f"Extraction de {filename}...")
        content = z.read(filename)
        dbutils.fs.put(
            f"/Volumes/transport/bronze/gtfs_static/{filename.split('.')[0]}-{runid}.txt",
            content.decode("utf-8"),
            overwrite=True
        )

print("✅ Done !")
```
