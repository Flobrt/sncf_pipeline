```python
df_stop = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("/Volumes/transport/bronze/gtfs_static/night_stops.csv")
)

df_trip = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("/Volumes/transport/bronze/gtfs_static/night_trips.csv")
)

df_stop_times = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("/Volumes/transport/bronze/gtfs_static/night_trips_stop.csv")
)

df_route = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("/Volumes/transport/bronze/gtfs_static/night_routes.csv")
)

df_calendar = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("/Volumes/transport/bronze/gtfs_static/night_calendar_dates.csv")
)
```

```python
df_stop = df_stop.select("stop_id", "stop_name", "stop_lat", "stop_lon")
df_trip = df_trip.select("trip_id", "route_id", "service_id")
df_stop_times = df_stop_times.select("stop_id", "trip_id", "arrival_time", "departure_time", "stop_sequence")
df_route = df_route.select("route_id", "route_short_name", "route_long_name")
```

```python
df_stop.write.mode("overwrite").saveAsTable("transport.silver.gtfs_stops_night")
df_trip.write.mode("overwrite").saveAsTable("transport.silver.gtfs_trips_night")
df_stop_times.write.mode("overwrite").saveAsTable("transport.silver.gtfs_stop_times_night")
df_route.write.mode("overwrite").saveAsTable("transport.silver.gtfs_routes_night")
df_calendar.write.mode("overwrite").saveAsTable("transport.silver.gtfs_calendar_dates_night")
```
