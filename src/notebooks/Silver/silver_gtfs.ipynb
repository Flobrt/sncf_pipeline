```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("GTFS Silver").getOrCreate()
```

```python
runid = "2026-05-13_12-43-46"
# runid = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
```

# READ FILES

```python
df_stop = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("../../../data/bronze/gtfs_static/stops.txt")
)

df_trip = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("../../../data/bronze/gtfs_static/trips.txt")
)

df_stop_times = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("../../../data/bronze/gtfs_static/stop_times.txt")
)

df_route = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("../../../data/bronze/gtfs_static/routes.txt")
)

df_calendar = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("../../../data/bronze/gtfs_static/calendar_dates.txt")
)
```

```python
df_stop = df_stop.select("stop_id", "stop_name", "stop_lat", "stop_lon")
df_trip = df_trip.select("trip_id", "route_id", "service_id")
df_stop_times = df_stop_times.select("stop_id", "trip_id", "arrival_time", "departure_time", "stop_sequence")
df_route = df_route.select("route_id", "route_short_name", "route_long_name")
```

```python
df_stop.write.mode("overwrite").parquet(f"../../../data/silver/gtfs_static/stops")
df_trip.write.mode("overwrite").parquet(f"../../../data/silver/gtfs_static/trips")
df_stop_times.write.mode("overwrite").parquet(f"../../../data/silver/gtfs_static/stop_times")
df_route.write.mode("overwrite").parquet(f"../../../data/silver/gtfs_static/routes")
df_calendar.write.mode("overwrite").parquet(f"../../../data/silver/gtfs_static/calendar_dates")
```
